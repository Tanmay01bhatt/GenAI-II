# WorkFlow Orchestration

### Sequential Agent 

- A SequentialAgent executes a list of sub-agents or tools one after another, in a strict, linear order.

```python
from google.adk.agents.sequential_agent import SequentialAgent

# The root_agent is a workflow that orchestrates other agents
root_agent = SequentialAgent(
    agents=[sub_agent_1, sub_agent_2, sub_agent_3]
)
```

- agents: (required) This parameter takes a Python list of agent instances. The order of the agents in this list is critical, as it defines the exact, unchangeable sequence of execution.

### ParallelAgent

- A ParallelAgent executes multiple sub-agents or tools concurrently. This agent is designed for optimization and efficiency, greatly speeding up workflows where tasks can be performed independently.

```python
from google.adk.agents.parallel_agent import ParallelAgent

# Define the ParallelAgent to manage the concurrent execution
root_agent = ParallelAgent(
    agents=[independent_task_agent_A, independent_task_agent_B]
)
```

- agents: (required) This parameter takes a Python list of agent instances. Unlike with the SequentialAgent, the order of agents in this list does not imply an execution sequence, as they will all be started concurrently.

### LoopAgent

- A LoopAgent repeatedly executes its sub-agents in a loop. It is designed for workflows that require iteration, repetition, or refinement.

- The sequence of agents inside the loop is executed sequentially, just like in a SequentialAgent. Once the sequence completes, the LoopAgent checks if a termination condition has been met

```python
from google.adk.agents.loop_agent import LoopAgent

# Define the LoopAgent itself
repeating_process = LoopAgent(
    agents=[iterative_task_agent_A, iterative_task_agent_B],
    max_iterations=5
)
```

- agents: (required) It is a list of agent instances that form the body of the loop. They will be executed sequentially in each iteration.

-max_iterations: (optional) It is an integer that sets a maximum number of times the loop can run.
